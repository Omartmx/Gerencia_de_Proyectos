#DOMINIO DE GOBERNANZA

El "Dominio de Gobernanza" en la gerencia de proyectos es el sistema integral que define cómo se dirige, controla y supervisa un proyecto para asegurar que cumpla con sus objetivos estratégicos. La imagen del edificio en construcción representa que la gobernanza es la estructura base sobre la cual se construye el éxito del proyecto, utilizando diversas herramientas y planos.

Se divide en 6 pilares fundamentales para el director de proyecto (Project Manager):

Marco de Gobernanza: Son las políticas, reglas y la estructura organizacional que definen cómo se alineará el proyecto con la estrategia de la empresa. Responde a la pregunta: ¿Bajo qué reglas nos regimos?

Roles Clave: Identifica a los responsables de la toma de decisiones y la supervisión (como el Project Manager, el Project Sponsor o el PMO). Define claramente quién hace qué.

Toma de Decisiones: Establece los procesos, umbrales y la autoridad para tomar decisiones críticas (por ejemplo, aprobación de presupuestos, cambios en el alcance). Evita cuellos de botella y agiliza la gestión.

Autogobierno Ágil: Un concepto moderno donde los equipos de trabajo (mencionados como "Agile Scrum Teams" en la imagen) tienen la autonomía y madurez para tomar decisiones sobre su propio trabajo diario, siempre alineados con el marco general.

Plan de Gobernanza: Es el documento o estrategia formal que detalla cómo se implementará todo lo anterior. Es la hoja de ruta de la supervisión del proyecto.

Matriz RACI: Es la herramienta práctica de gestión que clarifica las responsabilidades de cada miembro del equipo en cada tarea o entregable (quién es Responsable, Aprobador, Consultado e Informado).

Conclusión para la Gerencia de Proyectos:
Este dominio asegura que el proyecto no sea un caos, sino un esfuerzo organizado. Establece un equilibrio entre el control (marcos, planes, matrices) y la flexibilidad (autogobierno ágil), garantizando que el proyecto se entregue con éxito, dentro del presupuesto y cumpliendo con las expectativas de los interesados (stakeholders).
