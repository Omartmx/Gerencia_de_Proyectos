Resumen: Ruta Crítica y Compresión de Proyectos

La Ruta Crítica y Compresión son técnicas de gestión de proyectos que permiten planificar, controlar y reducir el tiempo de ejecución de un proyecto. Se utilizan principalmente para saber qué actividades son más importantes, cuánto puede retrasarse una actividad y cómo terminar el proyecto en menos tiempo.

1.  Red lógica del proyecto

La red lógica representa gráficamente las actividades de un proyecto y la relación que existe entre ellas.

Cada actividad tiene una duración y unas dependencias, es decir, algunas actividades deben terminar antes de que otras puedan comenzar.

Ejemplo

Supongamos un proyecto con estas actividades:

Actividad	Duración	Predecesora
A: Planificación	3 días	—
B: Diseño	4 días	A
C: Compra de materiales	2 días	A
D: Construcción	5 días	B y C
E: Pruebas	2 días	D

La red sería aproximadamente:

A → B → D → E
A → C → D → E

Esto permite visualizar qué actividades dependen de otras y determinar cuánto tiempo puede durar todo el proyecto.

2.  Ruta crítica

La ruta crítica es la secuencia de actividades que determina la duración mínima total del proyecto.

Las actividades que pertenecen a esta ruta normalmente tienen holgura igual a cero.

Siguiendo el ejemplo:

A = 3 días
B = 4 días
C = 2 días
D = 5 días
E = 2 días

Tenemos dos caminos:

A → B → D → E

3 + 4 + 5 + 2 = 14 días

A → C → D → E

3 + 2 + 5 + 2 = 12 días

Por lo tanto:

Ruta crítica = A → B → D → E

Y la duración del proyecto es de 14 días.

 Si una actividad de la ruta crítica se retrasa, normalmente se retrasa todo el proyecto, salvo que se tomen medidas para recuperar el tiempo.

3.  Holgura

La holgura indica cuánto puede retrasarse una actividad sin retrasar la fecha final del proyecto.

Por ejemplo:

Una actividad tiene una holgura de 0 días → es crítica.
Una actividad tiene una holgura de 2 días → puede retrasarse hasta 2 días sin afectar la fecha final.
Tipos de tiempos

Para calcular la ruta crítica se suelen utilizar:

ES (Early Start): inicio más temprano.
EF (Early Finish): finalización más temprana.
LS (Late Start): inicio más tardío.
LF (Late Finish): finalización más tardía.

La holgura puede calcularse como:

Holgura = LS − ES

o:

Holgura = LF − EF

Si:

Holgura = 0

la actividad pertenece a la ruta crítica.

4.  ¿Por qué es importante la ruta crítica?

Permite al administrador del proyecto:

Identificar las actividades que no pueden retrasarse.
Conocer la duración mínima del proyecto.
Priorizar recursos.
Detectar posibles retrasos.
Tomar decisiones para reducir el tiempo.
Controlar mejor el cronograma.

En otras palabras, la ruta crítica responde principalmente a:

¿Qué actividades determinan cuándo puedo terminar el proyecto?

5.  Compresión del proyecto

La compresión del proyecto consiste en intentar reducir la duración del proyecto sin cambiar significativamente su alcance.

Se utiliza cuando, por ejemplo:

El cliente necesita el proyecto antes.
Existe un retraso.
Se debe cumplir una fecha límite.
Se quiere entregar antes de lo planeado.

Existen principalmente dos técnicas:

1. Crashing

Consiste en agregar recursos para reducir la duración de determinadas actividades.

Por ejemplo:

Una actividad normalmente tarda:

5 días → 2 trabajadores

Podemos contratar un trabajador adicional y conseguir:

3 días → 3 trabajadores

Sin embargo, esto generalmente aumenta los costos.

Ejemplos de recursos adicionales:

Más trabajadores.
Horas extras.
Equipos adicionales.
Contratación externa.
Materiales adicionales.
Ventaja

Puede reducir considerablemente el tiempo.

Desventaja

 Aumenta los costos.

6.  Fast Tracking

El Fast Tracking consiste en realizar actividades en paralelo que originalmente estaban programadas una después de otra.

Por ejemplo:

Plan original

Diseño → Construcción → Pruebas

Si algunas actividades del diseño ya están terminadas, podemos comenzar parte de la construcción antes de finalizar completamente el diseño:

Diseño ──────┐
** ├→ Construcción → Pruebas**
Diseño parcial┘

Esto permite ahorrar tiempo.

Ventaja

No necesariamente requiere contratar más recursos.

Desventaja

 Aumenta el riesgo de errores y retrabajo porque algunas actividades se realizan simultáneamente.

7.  Crashing vs. Fast Tracking
Característica	Crashing	Fast Tracking
¿Qué hace?	Agrega recursos	Superpone actividades
Reduce tiempo	Sí	Sí
Aumenta costos	Generalmente sí	No necesariamente
Aumenta riesgos	Moderadamente	Sí, considerablemente
Ejemplo	Contratar más trabajadores	Trabajar diseño y construcción parcialmente al mismo tiempo
8.  Cronograma optimizado

Un cronograma optimizado es aquel que busca encontrar un equilibrio entre:

Tiempo + Costo + Recursos + Riesgo

No siempre significa terminar el proyecto lo más rápido posible.

Por ejemplo:

Proyecto original: 30 días y $10.000

Después de analizar la ruta crítica:

Proyecto comprimido: 25 días y $12.000

La empresa debe decidir si los 5 días ahorrados justifican los $2.000 adicionales.

Por eso, la optimización consiste en encontrar una alternativa que cumpla los objetivos del proyecto de manera eficiente.

9. Proceso para optimizar un proyecto

Un proceso sencillo sería:

1. Identificar actividades
⬇️
2. Establecer dependencias
⬇️
3. Construir la red lógica
⬇️
4. Calcular la ruta crítica
⬇️
5. Calcular las holguras
⬇️
6. Identificar actividades que pueden comprimirse
⬇️
7. Aplicar Crashing o Fast Tracking
⬇️
8. Evaluar costo y riesgo
⬇️
9. Crear el cronograma optimizado

Ejemplo sencillo de compresión

Supongamos que la ruta crítica es:

A → B → C → D

Duración:

4 + 6 + 5 + 3 = 18 días

Queremos terminar en 15 días.

Podemos analizar qué actividades pueden reducirse:

Actividad	Normal	Comprimida	Reducción
A	4	3	1 día
B	6	4	2 días
C	5	4	1 día
D	3	3	0 días

Si comprimimos B y C:

4 + 4 + 4 + 3 = 15 días

El proyecto pasa de 18 a 15 días.

Pero debemos comprobar cuánto cuesta realizar esa compresión y qué riesgos genera.
