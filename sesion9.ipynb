# Resumen Dominando El Cronograma

1. Dominio de la Programación (Schedule Management)

Es el área de conocimiento que define cómo se planifica, desarrolla, controla y cierra el cronograma de un proyecto. Incluye elaborar el Plan de Gestión del Cronograma: qué metodología se usará (ruta crítica, ágil, etc.), qué herramienta de software, y cómo se medirá y controlará el avance.

2. Definir Actividades

Es el proceso de descomponer los paquetes de trabajo del WBS/EDT en actividades específicas necesarias para producir los entregables. El resultado es la lista de actividades, con sus atributos (duración estimada, recursos necesarios, restricciones) y los hitos del proyecto.

3. Secuencias y Lógica

Consiste en ordenar las actividades según sus relaciones de dependencia, usando el Método de Diagramación por Precedencia (PDM). Las 4 relaciones lógicas son:

Fin-Inicio (FS): la más común; una tarea empieza cuando termina la anterior.
Inicio-Inicio (SS): ambas inician juntas.
Fin-Fin (FF): ambas terminan juntas.
Inicio-Fin (SF): poco usada.

También se identifican dependencias obligatorias, discrecionales, externas e internas. El resultado es el Diagrama de Red del Cronograma.

4. Adelantos y Retrasos (Leads & Lags)
Adelanto (Lead): permite que una actividad sucesora empiece antes de que termine la predecesora (ej. "-3 días").
Retraso (Lag): obliga a esperar un tiempo entre el fin de una actividad y el inicio de la siguiente (ej. tiempo de secado de pintura, "+2 días").

Estos ajustes afinan la lógica de secuenciación sin cambiar el tipo de relación.

5. Tutorial de Software

Aplicación práctica en herramientas de programación de proyectos —típicamente Microsoft Project, Primavera P6 u otras— para construir el cronograma: cargar actividades, definir predecesoras, asignar duraciones/recursos, calcular la ruta crítica y visualizar el diagrama de Gantt.

6. Entregables Clave

Los productos finales de este proceso son:

Lista de actividades y atributos
Diagrama de red del cronograma
Línea base del cronograma (Schedule Baseline)
Diagrama de Gantt
Cronograma del proyecto aprobado

En conjunto, este tema cubre el flujo completo de Gestión del Cronograma: desde definir qué se debe hacer, hasta ordenarlo lógicamente, ajustarlo con adelantos/retrasos, construirlo en una herramienta de software y obtener los entregables formales que sirven de línea base para controlar el proyecto — justo el tipo de estructura que usamos en el cronograma que armamos antes.
