Objetivo

Implementar una estructura de Gerencia de Proyectos en Jira para el proyecto "Desarrollo de Aplicación de IA", en el espacio "Mi equipo de gestión de proyectos" (proyecto KAN), sin enfocarse en código sino en el proceso de gestión.

 Paso 1: Conexión

Conectaste el conector Atlassian Rovo, lo que me permitió crear y editar contenido directamente en tu Jira.

 Paso 2: Estructura de fases (5 Épicas)

Se crearon 5 Épicas basadas en las fases estándar de gerencia de proyectos (PMI):

Épica	Fase  
KAN-4	Inicio del Proyecto  
KAN-5	Planificación  
KAN-6	Ejecución  
KAN-7	Monitoreo y Control  
KAN-8	Cierre  
 Paso 3: 20 tareas distribuidas

Cada épica recibió tareas concretas y ligadas como hijas:

Inicio (3): Acta de constitución, stakeholders,
alcance preliminar      
Planificación (6): EDT/WBS, cronograma, presupuesto, riesgos, comunicaciones, calidad
Ejecución (4): Conformar equipo, requerimientos del modelo IA, reuniones de seguimiento, coordinación de entregables        
Monitoreo y Control (4): KPIs, control de cambios, riesgos activos, reportes de estado  
Cierre (3): Aceptación del cliente, lecciones aprendidas, cierre administrativo  

 Paso 4: Fechas y prioridades
Duración total: 1 mes (14 sept – 14 oct 2026)
Fechas límite escalonadas por fase, con Monitoreo corriendo en paralelo a Ejecución
Prioridades Alta/Media asignadas según criticidad de cada tarea  
 Paso 5: Asignación

Las 20 tareas quedaron asignadas a omar peña (tú), como responsable único mientras se define el resto del equipo.
