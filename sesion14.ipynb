1.  Introducción a la Sastrería

La sastrería (Tailoring) consiste en adaptar la forma de gestionar un proyecto a sus características particulares, en lugar de aplicar exactamente la misma metodología a todos los proyectos.

En otras palabras:

No todos los proyectos necesitan el mismo nivel de procesos, documentación, controles o herramientas.

El PMBOK® 8 dedica un capítulo específico a la sastrería y establece cuatro pasos principales:

Seleccionar el enfoque inicial de desarrollo.
Adaptarlo a la organización.
Adaptarlo al proyecto.
Implementar la mejora continua.

Por ejemplo, un proyecto de desarrollo de software puede utilizar un enfoque adaptativo/ágil, mientras que un proyecto de construcción puede requerir un enfoque más predictivo. También puede utilizarse un enfoque híbrido. El PMBOK® 8 contempla explícitamente los enfoques predictivo, adaptativo e híbrido.

2.  ¿Qué se puede adaptar?

La sastrería puede aplicarse a diferentes elementos del proyecto, entre ellos:

Ciclo de vida del proyecto
Enfoque de desarrollo
Procesos
Nivel de documentación
Herramientas y técnicas
Forma de involucrar a los interesados
Controles y revisiones
Métodos de comunicación

El objetivo es encontrar un equilibrio: ni gestionar el proyecto con demasiada burocracia ni hacerlo sin los controles necesarios.

El PMBOK® 8 señala que la sastrería debe considerar tanto el contexto de la organización como las características específicas del proyecto.

Ejemplo

Proyecto pequeño de una aplicación web:

Equipo de 4 personas.
Requisitos relativamente simples.
Entregas frecuentes.

Podría utilizarse un enfoque adaptativo y una documentación ligera.

Proyecto grande de infraestructura:

Muchos interesados.
Contratos.
Riesgos importantes.
Requisitos regulatorios.

Podría necesitar mayor gobernanza, documentación, controles y un enfoque predictivo o híbrido.

3.  Enfoques de desarrollo

Según PMBOK® 8, existen tres enfoques principales:

Enfoque	Característica
Predictivo	Se define gran parte del alcance antes de ejecutar el trabajo.
Adaptativo	Se trabaja de manera iterativa e incremental, permitiendo cambios frecuentes.
Híbrido	Combina elementos predictivos y adaptativos.

La selección depende de factores como el producto, los entregables, el proyecto y la organización.

4.  Actividades de Cierre

El cierre es la etapa en la que se confirma que el proyecto o una fase puede finalizar formalmente.

En PMBOK® 8, Closing aparece como uno de los cinco Focus Areas:

Initiating
Planning
Executing
Monitoring and Controlling
Closing

El cierre no significa simplemente decir “ya terminamos”. Implica comprobar que el trabajo realizado puede ser aceptado y que la organización puede pasar de la ejecución del proyecto a la operación o siguiente etapa.

Actividades importantes de cierre
Verificar que los entregables estén completos.
Obtener la aceptación formal correspondiente.
Confirmar que se cumplieron los objetivos y criterios establecidos.
Finalizar contratos y adquisiciones cuando corresponda.
Transferir los entregables a los usuarios u operación.
Documentar información importante del proyecto.
Registrar lecciones aprendidas.
Liberar los recursos del proyecto.
Cerrar formalmente el proyecto o fase.

Estas actividades son coherentes con la práctica de cierre descrita por PMI y con el enfoque de valor y resultados del PMBOK®.

5.  Taller: Retrospectiva

La retrospectiva es especialmente importante en proyectos con enfoques adaptativos, aunque la idea de aprendizaje puede utilizarse en cualquier proyecto.

Su objetivo es responder preguntas como:

¿Qué salió bien?

Identificar prácticas que deberían mantenerse.

¿Qué salió mal?

Identificar problemas, obstáculos o situaciones que afectaron el proyecto.

¿Qué podemos mejorar?

Determinar acciones concretas para futuros trabajos.

¿Qué aprendimos?

Convertir la experiencia del proyecto en conocimiento organizacional.

Esto se relaciona directamente con la mejora continua, que el PMBOK® 8 incluye como parte del proceso de sastrería.

6.  Entregables finales

Antes de cerrar un proyecto se deben revisar los entregables finales.

Una lista sencilla sería:

Entregable → Verificación → Aceptación → Transferencia → Cierre

Por ejemplo:

Desarrollo de una aplicación
↓
Pruebas finales
↓
Corrección de errores
↓
Aceptación del cliente
↓
Entrega de documentación y aplicación
↓
Cierre del proyecto

Esto permite relacionar el cierre con uno de los conceptos centrales del PMBOK® 8: la entrega de valor, no solamente completar actividades. PMI señala que la octava edición pone énfasis en conectar los proyectos con el valor organizacional.
