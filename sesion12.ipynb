Resumen: Monitoreo y Cambio  
El monitoreo y control es el proceso mediante el cual se rastrea, revisa y regula el progreso del proyecto para identificar áreas que requieren cambios y ejecutar las acciones correctivas necesarias . En el contexto del PMBOK 8, este grupo de procesos se articula como un puente entre la planificación y la entrega de valor, asegurando que el proyecto se mantenga alineado con sus objetivos estratégicos .

Monitoreo y Control: Este proceso, también llamado "Monitorear y Controlar el Trabajo del Proyecto", implica recopilar, medir y distribuir información sobre el desempeño, comparando los resultados reales con el plan para evaluar tendencias y detectar desviaciones .

Enfoque en datos: El PMBOK 8 enfatiza la toma de decisiones basada en datos. Se utilizan indicadores clave de desempeño (KPI) para evaluar el avance en tiempo, costo, alcance y calidad .

Cambio Inesperado: Los cambios en los proyectos son inevitables, y el PMBOK 8 reconoce que la flexibilidad es una habilidad gerencial esencial. Un "cambio inesperado" no debe gestionarse de forma reactiva, sino mediante una adaptación estructurada que evalúe el impacto en el valor del proyecto .

Acciones correctivas: Se deben determinar acciones preventivas o correctivas sobre las actividades del plan, monitoreando su efectividad para resolver problemas de desempeño .

Comité de Control: El PMBOK 8 resalta la importancia de la Gobernanza como dominio de desempeño, que establece las reglas, marcos y procesos de toma de decisiones del proyecto . El Comité de Control actúa como el órgano que supervisa las solicitudes de cambio, aprueba las que sean convenientes y gestiona su implementación, asegurando la alineación con los objetivos estratégicos de la organización .

Tutorial Práctico: La octava edición del PMBOK reintroduce los procesos y las ITTO (Entradas, Herramientas, Técnicas y Salidas) como una guía práctica no prescriptiva .

Herramientas clave: El Control Integrado de Cambios es la herramienta fundamental para gestionar las modificaciones. Se recomienda el uso del ciclo PDCA (Planificar-Hacer-Verificar-Actuar) para evaluar el impacto de un cambio antes de adoptarlo o abandonarlo, evitando consecuencias no deseadas .

Tailoring (Adaptación): La guía no debe aplicarse de manera rígida; el gerente del proyecto debe adaptar las prácticas y herramientas según el contexto, la complejidad y el enfoque de entrega (predictivo, ágil o híbrido) .

Entregables Clave: La gestión del cambio debe producir actualizaciones formales a los documentos y planes del proyecto .

Integración con la IA: En línea con el PMBOK 8, se espera que el gerente de proyecto aproveche herramientas de Inteligencia Artificial para automatizar el análisis de datos, la predicción de riesgos y la generación de informes, liberando tiempo para el liderazgo y la toma de decisiones estratégicas .

En resumen, la sesión aborda cómo el enfoque en el valor y la adaptabilidad permiten gestionar el desempeño del proyecto, utilizando estructuras de gobernanza sólidas y herramientas prácticas para transformar los cambios inesperados en oportunidades de mejora .
