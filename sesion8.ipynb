# EDT y EDV

EDT — Estructura de Desglose del Trabajo

La EDT es una herramienta de gestión de proyectos que sirve para dividir un proyecto grande en partes más pequeñas y fáciles de organizar.

Su objetivo es responder:

¿Qué trabajo debemos realizar para completar el proyecto?

La EDT comienza con el proyecto completo y se va dividiendo en:

Proyecto → Fases → Entregables → Paquetes de trabajo

Ejemplo

Proyecto: Desarrollar una aplicación móvil

1. Aplicación móvil  
1.1 Análisis  
1.1.1 Requisitos  
1.1.2 Análisis de usuarios  
1.2 Diseño  
1.2.1 Diseño de interfaz  
1.2.2 Diseño de base de datos  
1.3 Desarrollo  
1.3.1 Programación  
1.3.2 Integración  
1.4 Pruebas  
1.4.1 Pruebas funcionales  
1.4.2 Corrección de errores

 La EDT descompone el trabajo, pero no necesariamente indica cuándo se realizará.

 EDV — Estructura de Desglose de Valor

La EDV organiza el proyecto desde el punto de vista del valor que se espera obtener.

Su objetivo es responder:

¿Qué valor o beneficios queremos obtener con el proyecto?

Permite relacionar las actividades y entregables con los beneficios y resultados esperados.

Ejemplo

Proyecto: Desarrollar una aplicación móvil para ventas

Proyecto  
Aumentar las ventas  
Aplicación móvil  
Catálogo de productos  
Sistema de pedidos  
Mejorar la atención al cliente  
Notificaciones  
Seguimiento de pedidos  
Reducir costos  
Automatización de procesos  
Reportes digitales  


 La EDV se enfoca en el valor y los beneficios, mientras que la EDT se enfoca en el trabajo que debe hacerse.

 Diferencia entre EDT y EDV
EDT	EDV
Estructura el trabajo	Estructura el valor
¿Qué debemos hacer?	¿Qué valor obtenemos?
Se enfoca en actividades y entregables	Se enfoca en beneficios y resultados
Ayuda a planificar el alcance	Ayuda a identificar beneficios
Divide el proyecto en partes pequeñas	Relaciona resultados con valor
 Para memorizar

EDT = Trabajo que hay que hacer.
EDV = Valor que se quiere obtener.

Ejemplo rápido:
Si haces una aplicación:

EDT: diseñar → programar → probar → implementar.
EDV: aumentar ventas → mejorar servicio → reducir costos.
