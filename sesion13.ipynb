La Gestión del Valor Ganado (EVM) es una metodología que integra el alcance, el cronograma y los costos para medir objetivamente el desempeño del proyecto y predecir su futuro . Se basa en la comparación de tres valores fundamentales y sus métricas derivadas, lo que permite identificar desviaciones tempranas y tomar decisiones correctivas basadas en datos .

Los Tres Valores Fundamentales
Para aplicar EVM es necesario establecer primero una Línea Base de Medición del Desempeño (PMB) y luego medir tres variables clave :

Valor Planificado (PV): Es el presupuesto autorizado asignado al trabajo que se planeó realizar en un momento específico. Responde a la pregunta: ¿Cuánto trabajo deberíamos haber hecho ya? .

Valor Ganado (EV): Es el valor del trabajo que realmente se ha completado, expresado en términos del presupuesto aprobado para ese trabajo (ej. % de avance × presupuesto de la tarea). Responde: ¿Cuánto trabajo hemos hecho realmente? .

Costo Real (AC): Es el costo total en el que se ha incurrido para llevar a cabo el trabajo del EV. Responde: ¿Cuánto nos ha costado el trabajo realizado? .

 Métricas de Desempeño y Pronóstico
A partir de estos tres valores se calculan indicadores que revelan la salud del proyecto :

Métrica	Fórmula	Interpretación
Variación del Costo (CV)	CV = EV - AC	Mide la desviación presupuestaria en un momento dado. Positiva indica ahorro; negativa indica sobrecosto .  

Variación del Cronograma (SV)	SV = EV - PV	Mide la desviación del cronograma. Positiva indica adelanto; negativa indica retraso .

Índice de Desempeño del Costo (CPI)	CPI = EV / AC	Mide la eficiencia del costo. > 1.0 es bueno (gastamos menos de lo previsto); < 1.0 indica sobrecosto .

Índice de Desempeño del Cronograma (SPI)	SPI = EV / PV	Mide la eficiencia del cronograma. > 1.0 es bueno (avanzamos más rápido); < 1.0 indica retraso .

Estimación a la Conclusión (EAC)	EAC = BAC / CPI	Pronostica el costo total final del proyecto asumiendo que se mantiene la eficiencia actual .

Índice de Desempeño para Terminar (TCPI)	TCPI = (BAC - EV) / (BAC - AC)	Indica la eficiencia (CPI) que se necesita de aquí en adelante para cumplir con el presupuesto original (BAC) .

 Diagnóstico y Acción
El análisis combinado de CPI y SPI permite un diagnóstico rápido y guía las acciones a tomar :

CPI > 1.0 y SPI > 1.0: Proyecto saludable (bajo presupuesto y adelantado).

CPI < 1.0 y SPI < 1.0: Situación crítica (sobrecosto y retrasado). Requiere acción inmediata para revisar el alcance, los recursos o el presupuesto .

CPI < 1.0 y SPI > 1.0: Se gasta más de lo previsto, pero se avanza rápido. Puede indicar un uso ineficiente de recursos o una aceleración innecesaria .

CPI > 1.0 y SPI < 1.0: Se está dentro del presupuesto, pero el proyecto va retrasado. Sugiere evaluar la asignación de recursos o la ruta crítica del cronograma .

Para implementar EVM correctamente, los proyectos deben tener un alcance claramente definido (a través de la EDT), responsabilidades asignadas, un cronograma detallado y un presupuesto distribuido . La medición del progreso debe ser objetiva y realizarse de forma periódica y consistente .

Los 6 Pasos de la Gestión de Riesgos
El proceso de gestión de riesgos se estructura en seis fases que abarcan desde la planificación inicial hasta el monitoreo continuo.

Planificar la Gestión de Riesgos: Define cómo se estructurarán y ejecutarán las actividades de gestión de riesgos. Incluye metodología, roles, responsabilidades, presupuesto y calendario de actividades.

Identificar los Riesgos: Proceso de determinar qué riesgos podrían afectar al proyecto y documentar sus características. Se utilizan técnicas como tormenta de ideas, análisis FODA, entrevistas y listas de verificación.

Análisis Cualitativo: Evalúa y prioriza los riesgos según su probabilidad de ocurrencia e impacto en los objetivos. Se emplea la matriz de probabilidad e impacto para clasificar los riesgos en categorías como alto, medio o bajo.

Análisis Cuantitativo: Aplica métodos matemáticos para asignar valores numéricos a los riesgos. Incluye Simulación de Monte Carlo (analiza miles de escenarios posibles), Árboles de Decisión (determinan la mejor decisión considerando incertidumbre) y Análisis de Sensibilidad (identifica qué variables tienen mayor impacto).

Planificar la Respuesta a los Riesgos: Define las acciones para abordar los riesgos según su tipo (amenaza u oportunidad).

Implementar y Monitorear: Ejecuta los planes de respuesta y realiza seguimiento continuo, reevaluando riesgos y aplicando auditorías periódicas.
