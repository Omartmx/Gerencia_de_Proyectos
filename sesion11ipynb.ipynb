Dominio Financiero — Resumen del contenido

1. Valor del Proyecto
Se enfoca en cómo evaluar si un proyecto genera valor económico, normalmente usando herramientas como el Valor Presente Neto (VPN/NPV), la Tasa Interna de Retorno (TIR) y el análisis costo-beneficio.

2. CapEx vs OpEx
Diferencia entre gastos de capital (CapEx — inversiones en activos de largo plazo, como equipos o infraestructura) y gastos operativos (OpEx — costos recurrentes del día a día, como salarios o suministros). Es clave entender esta distinción porque afecta cómo se contabilizan, presupuestan y depreciar los gastos.

3. Métricas de Viabilidad
Indicadores financieros usados para determinar si un proyecto o inversión es viable: ROI (retorno sobre la inversión), periodo de recuperación (payback period), punto de equilibrio, entre otros.

4. Modelo de Flujo de Caja
Construcción de proyecciones de entradas y salidas de efectivo a lo largo del tiempo, esencial para anticipar necesidades de liquidez y sostenibilidad financiera del proyecto.

5. Entregables Clave
Los documentos o resultados finales esperados del análisis financiero — típicamente informes, modelos financieros, presentaciones ejecutivas o recomendaciones de inversión.
